```python
import os
import json
import numpy as np
import matplotlib.pyplot as plt
logs = [
    '/data/models/bin/2026_pathway_generation/grpo_logs/resnet18_20260218_230403/',
    '/data/models/bin/2026_pathway_generation/grpo_logs_steps_multiplicative/resnet18_20260221_201647',
    '/data/models/bin/2026_pathway_generation/grpo_logs_steps_multiplicative/resnet18_20260221_202209',
    '/data/models/bin/2026_pathway_generation/grpo_logs_steps_additive/resnet18_20260221_142852',
    '/data/models/bin/2026_pathway_generation/grpo_logs_steps_additive/resnet18_20260221_203455',
]
labels = ['resnet18', 'resnet18_multiplicative (0.1)', 'resnet18_multiplicative (0.2)', 'resnet18_additive (0.1)', 'resnet18_additive (0.2)']
fig, axs = plt.subplots(1,2, figsize=(10,4))
for i, log in enumerate(logs):
    with open(log + '/history.json', 'r') as f:
        history = json.load(f)
    print(labels[i])
    print('Train IOU:', history['train_iou'])
    print('Val IOU:', history['val_iou'])
    axs[0].plot(history[f'train_iou'], label=labels[i])
    axs[0].set_title('Training IOU')
    axs[0].set_xlabel('Epoch')
    axs[1].plot(history[f'val_iou'], label=labels[i])
    axs[1].set_title('Validation IOU')
    axs[1].set_xlabel('Epoch')
    plt.legend()
```
